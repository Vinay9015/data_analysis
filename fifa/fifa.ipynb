```python
columns=df.columns
n_col=[]
```

```python
#null columns
for i in columns:
    if df[i].isnull().any():
        n_col.append(i)
n_col
```

```python
#null percentage
for i in n_col:
    n_p=(df[i].isnull().sum()/len(df[i]))*100
    print(i,"null percentage is :",n_p)
```
